```python
import os
import datetime as dt


cliFolder = '/content/drive/My Drive/Colab Notebooks/TeamProjects/Merilynn/output'
outputFolder = '/content/drive/My Drive/Colab Notebooks/TeamProjects/Merilynn/formatted_output'

cliFiles = os.listdir(cliFolder)


START_DATE = '1981,01,01'
END_DATE = '2025,12,31'


time = dt.datetime.strptime(START_DATE, '%Y,%m,%d')
time_end = dt.datetime.strptime(END_DATE, '%Y,%m,%d')
dates = [time]

while time < time_end:
  time = time + dt.timedelta(seconds=86400)
  dates.append(time)

print(dates[0])
print(dates[-1])

header = 'i,Year,Month,Day,Precip(mm),Max(C),Min(C),Dew(C),Solrad(W/m2),Wind(m/s)'

for i, f in enumerate(cliFiles):

  cli_fpath = os.path.join(cliFolder, cliFiles[i])
  with open(cli_fpath) as f_cli:
    lines = f_cli.readlines()

  w, p, tx, tm, sr, w, td   = [], [], [], [], [], [], []

  for line in lines[15:-1]:
    p.append(line[12:17])
    tx.append(line[36:41])
    tm.append(line[42:47])
    td.append(line[64:69])
    sr.append(line[48:52])
    w.append(line[53:57])

  out_fpath = os.path.join(outputFolder, cliFiles[i][:-4] + '.csv')
  with open(out_fpath, 'w') as f_out:
    f_out.write(header + '\n')
    for j, line in enumerate(lines[15:-1]):

      index = str(j)
      year = str(dates[j].year)
      month = str(dates[j].month)
      day = str(dates[j].day)
      precip = str(round(float(p[j]), 2))
      tmax = str(round(float(tx[j]), 2))
      tmin = str(round(float(tm[j]), 2))
      tdew = str(round(float(td[j]), 2))
      srad = str(round(float(sr[j])*41868.0/86400.0, 2))
      wind = w[j].lstrip()
      row = ','.join([index, year, month, day, precip, tmax, tmin, tdew, srad, wind])
      f_out.write(row + '\n')

```

```python
import os
import pandas as pd

folder = '/content/drive/My Drive/Colab Notebooks/TeamProjects/Merilynn/formatted_output'

waco_fpath = os.path.join(folder, 'tx419419.csv')
marlin_fpath = os.path.join(folder, 'tx415611.csv')

header = 'i,Year,Month,Day,Precip(mm),Max(C),Min(C),Dew(C),Solrad(W/m2),Wind(m/s)'
hdr_row = header.split(',')

with open(waco_fpath) as f_waco:
  waco_lines = f_waco.readlines()[1:]

with open(marlin_fpath) as f_marlin:
  marlin_lines = f_marlin.readlines()[1:]

print(waco_lines[0])

waco = []
for l in waco_lines:
  row = l.strip('\n').split(',')
  waco.append(row)

marlin = []
for l in marlin_lines:
  row = l.strip('\n').split(',')
  marlin.append(row)

waco_df = pd.DataFrame(data=waco, columns=hdr_row).astype(float)
marlin_df = pd.DataFrame(data=marlin, columns=hdr_row).astype(float)

print('precip sum: ', waco_df['Precip(mm)'].sum(), marlin_df['Precip(mm)'].sum())
print('max avg: ', waco_df['Max(C)'].mean(), marlin_df['Max(C)'].mean())
print('min avg: ', waco_df['Min(C)'].mean(), marlin_df['Min(C)'].mean())
print('dew avg: ', waco_df['Dew(C)'].mean(), marlin_df['Dew(C)'].mean())
print('srad avg: ', waco_df['Solrad(W/m2)'].mean(), marlin_df['Solrad(W/m2)'].mean())
print('wnd avg: ', waco_df['Wind(m/s)'].mean(), marlin_df['Wind(m/s)'].mean())

```
